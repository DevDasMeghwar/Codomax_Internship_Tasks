```python
avg_petal = df.groupby('species')['petal_length'].mean()
```

```python
avg_petal
```

```python
plt.bar(avg_petal.index, avg_petal.values)
plt.title('Average Petal Length by Species')
plt.xlabel('Species')
plt.ylabel('Average Petal Length (cm)')
```

```python
plt.plot(df.index[:30], df['sepal_length'][:30])
plt.title('Sepal Length Across First 30 Flowers')
plt.xlabel('Flower Index')
plt.ylabel('Sepal Length (cm)')

plt.show()
```

```python
species_count = df['species'].value_counts()
```

```python
species_count
```

```python
plt.pie(
    species_count.values,
    labels=species_count.index,
    autopct='%1.1f%%',
    startangle=90
)
plt.title('Distribution of Iris Species')

plt.show()
```
